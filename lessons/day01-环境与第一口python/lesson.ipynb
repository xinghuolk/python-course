# Day 01 · 环境与第一口 Python

> 你已有 Lua/C/JS 基础，所以重点不是“什么是变量”，而是 **Python 哪里和你熟悉的语言不一样**。

## 1. 环境与运行

- 用 `uv venv` 建虚拟环境，`uv sync` 装依赖。对比 C：Python 无需编译，解释执行。
- 三种运行方式：
  - **REPL**：终端敲 `python` 进交互式，即时验证想法。
  - **脚本**：`python foo.py` 或 `uv run foo.py`。
  - **Notebook**：VS Code 打开 `.ipynb`，逐格运行（就是你现在用的这种）。

## 2. 心智模型：一切皆对象

- 整数、函数、类本身都是对象，都有类型：`type(3)`、`type(print)` 都能问。
- 变量是“名字绑定到对象”，不是 C 那种“内存盒子”。赋值 `x = 5` 是把名字 `x` 绑到对象 `5`。
- `id(x)` 看对象身份（类似地址）。`a = b` 不复制对象，只是多一个名字指向它。

```
# 一切皆对象 + 名字绑定
print(type(3), type(print))
x = 5
print(id(x))
```

## 3. 标量类型

- `int`：**任意精度，永不溢出**（对比 C 的 32/64 位！`2 ** 100` 直接算）。
- `float`：双精度浮点。`bool`：`True`/`False`，而且 **bool 是 int 的子类**（`True + 1 == 2`）。
- `None`：空值，相当于别处的 null。`str`：字符串。
- 动态类型：变量不声明类型，类型属于对象不属于名字；但可写类型注解 `x: int = 3`（仅提示，不强制）。

```
# int 任意精度，永不溢出
print(2 ** 100)
print(True + 1)  # bool 是 int 子类
```

## 4. 数字运算

- `/` **永远是浮点除法**：`7 / 2 == 3.5`。整数除法用 `//`：`7 // 2 == 3`。
- `%` 取模，`**` 幂（`2 ** 10 == 1024`）。
- **没有 `++`/`--`**；自增用 `x += 1`。增强赋值 `+= -= *= //=` 都有。
- `divmod(a, b)` 一次返回 `(商, 余数)`，写时分秒转换很顺手。

```
# 除法与运算
print(7 / 2, 7 // 2, 7 % 2, 2 ** 10)
print(divmod(3661, 60))
```

## 5. 字符串

- **不可变**：不能 `s[0] = 'x'`，任何“修改”都返回新串。
- 引号：`'...'` 和 `"..."` 等价；三引号 `"""..."""` 写多行。
- f-string 首选：`f"{name} = {value}"`；调试神器 `f"{x=}"` 直接打印 `x=3`；补零 `f"{n:02d}"`。
- 切片 `s[start:stop:step]`：`s[1:4]`、负索引 `s[-1]`、反转 `s[::-1]`。
- 常用方法：`.lower() .upper() .strip() .split() .join() .replace()`，子串判断 `"a" in s`。

```
# 字符串：f-string、切片、方法
name, value = "x", 42
print(f"{name} = {value}")
print(f"{value=}")
s = "Hello, Python"
print(s[:5], s[-6:], s[::-1])
print("python" in s.lower())
```

## 6. 真值与布尔逻辑

- truthiness：空字符串/空列表/`0`/`None` 都为“假”，非空/非零为“真”。`if items:` 比 `if len(items) > 0:` 更 Pythonic。
- `and`/`or` **返回操作数本身**，不是布尔：`"" or "默认"` 得到 `"默认"`。
- `is` 比较是否同一对象，`==` 比较值。判空用 `x is None`，别用 `==`。
- **链式比较**：`0 < x < 10` 合法且直观（C/JS 里不能这么写）。

```
# 真值与布尔逻辑
print(bool(""), bool([]), bool(0), bool("hi"))
print("" or "默认值")
print(0 < 5 < 10)
```

## 7. 控制流

- 缩进即代码块，无 `{}`；冒号 `:` 起头。
- `if / elif / else`，没有 `switch`（新版有 `match`，以后再学）。
- 循环：`for x in 可迭代物`、`range(n)`、`while 条件`、`break`/`continue`。
- 特色：`for ... else`——循环正常结束（没 break）才执行 else，找东西时很有用。

```
# 控制流：for-in / range / for-else
for i in range(3):
    print(i, end=" ")
print()
for n in [3, 5, 7]:
    if n % 2 == 0:
        print("找到偶数", n)
        break
else:
    print("没有偶数")
```
